```python
# Customer Churn & Retention Analytics + GenAI

## Project Objective
The goal of this project is to analyze telecom customer churn,
identify the main churn drivers, find high-risk customer segments,
and recommend retention actions.

## Business Problem
The telecom company is losing customers due to churn.
Management wants to understand who is leaving, why they are leaving,
and which customer groups require retention attention.

## Tools Used
- Python
- Pandas
- PostgreSQL
- Power BI
- GenAI
```

```python
# Import pandas for data loading and data manipulation
import pandas as pd

# Import numpy for numerical operations
import numpy as np

# Import matplotlib for charts and visual analysis
import matplotlib.pyplot as plt

# Show all columns when displaying a DataFrame
pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")
```

```python
# Path of the raw telecom churn dataset
raw_file_path = r"D:\PROJECT\Customer_Churn_Retention_GenAI\Data\Raw\WA_Fn-UseC_-Telco-Customer-Churn.csv"

# Load the CSV file into a pandas DataFrame
df = pd.read_csv(raw_file_path)

# Show confirmation message
print("Dataset loaded successfully.")

# Display first 5 rows
df.head()
```

```python
# Check total rows and columns
print("Dataset Shape:")
print(df.shape)

# Display all column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check data types and non-null values
print("\nDataset Info:")
df.info()
```

```python
# Check missing values in each column
missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python
# Check blank string values in object/text columns
blank_values = (df.astype(str).apply(lambda col: col.str.strip() == "")).sum()

print("\nBlank Values in Each Column:")
print(blank_values)
```

```python
# Convert blank spaces in TotalCharges to NaN
df["TotalCharges"] = df["TotalCharges"].replace(" ", np.nan)

# Convert TotalCharges from text/object to numeric
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

# Check how many missing values are present after conversion
print("Missing values in TotalCharges:")
print(df["TotalCharges"].isnull().sum())

# Check the new data type
print("\nTotalCharges data type:")
print(df["TotalCharges"].dtype)
```

```python
# Check complete duplicate rows
duplicate_rows = df.duplicated().sum()

print("Total duplicate rows:")
print(duplicate_rows)

# Check duplicate customer IDs
duplicate_customer_ids = df["customerID"].duplicated().sum()

print("\nDuplicate customer IDs:")
print(duplicate_customer_ids)
```

```python
# List of important categorical columns
categorical_columns = [
    "gender",
    "SeniorCitizen",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod",
    "Churn"
]

# Display unique values for each categorical column
for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].value_counts(dropna=False))
```

```python
# Select important numeric columns
numeric_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

# Show summary statistics
print("Numeric Summary:")
print(df[numeric_columns].describe())
```

```python
# Check invalid or suspicious values

print("Tenure less than 0:")
print((df["tenure"] < 0).sum())

print("\nMonthlyCharges less than or equal to 0:")
print((df["MonthlyCharges"] <= 0).sum())

print("\nTotalCharges less than 0:")
print((df["TotalCharges"] < 0).sum())
```

```python
# Check how many TotalCharges values are missing before fixing
print("Missing TotalCharges before fix:")
print(df["TotalCharges"].isnull().sum())

# For new customers with tenure = 0,
# TotalCharges can reasonably be treated as 0
df.loc[
    (df["tenure"] == 0) & (df["TotalCharges"].isnull()),
    "TotalCharges"
] = 0

# Check again after fixing
print("\nMissing TotalCharges after fix:")
print(df["TotalCharges"].isnull().sum())
```

```python
# Create a numeric churn flag
# Churn = Yes  -> 1
# Churn = No   -> 0

df["Churn_Flag"] = np.where(
    df["Churn"] == "Yes",
    1,
    0
)

# Check the result
print(df[["Churn", "Churn_Flag"]].head(10))
```

```python
# Validate churn flag counts
print(df["Churn_Flag"].value_counts())
```

```python
# Create customer tenure groups
# This helps us compare churn across customer lifecycle stages

def create_tenure_group(tenure):
    if tenure <= 12:
        return "0-12 Months"
    elif tenure <= 24:
        return "13-24 Months"
    elif tenure <= 48:
        return "25-48 Months"
    else:
        return "49+ Months"

df["Tenure_Group"] = df["tenure"].apply(create_tenure_group)

# Check the new groups
print(df["Tenure_Group"].value_counts())
```

```python
# Create monthly charge bands
# This helps us compare churn across low, medium, and high paying customers

def create_charge_band(charge):
    if charge < 40:
        return "Low Charges"
    elif charge < 80:
        return "Medium Charges"
    else:
        return "High Charges"

df["Monthly_Charge_Band"] = df["MonthlyCharges"].apply(create_charge_band)

# Check the distribution
print(df["Monthly_Charge_Band"].value_counts())
```

```python
# Create customer value segments based on TotalCharges
# This helps us identify low, medium, and high-value customers

def create_value_segment(total_charge):
    if total_charge < 1000:
        return "Low Value"
    elif total_charge < 3000:
        return "Medium Value"
    else:
        return "High Value"

df["Customer_Value_Segment"] = df["TotalCharges"].apply(create_value_segment)

# Check the distribution
print(df["Customer_Value_Segment"].value_counts())
```

```python
# List of service-related columns
service_columns = [
    "PhoneService",
    "MultipleLines",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

# Count how many services are actively used by each customer
# Only exact "Yes" values are counted as active services
df["Service_Count"] = (
    df[service_columns]
    .eq("Yes")
    .sum(axis=1)
)

# Check the result
print(df[["customerID", "Service_Count"]].head(10))

# Check distribution
print("\nService Count Distribution:")
print(df["Service_Count"].value_counts().sort_index())
```

```python
# Create customer engagement segments based on number of active services
# This helps us classify customers by engagement level

def create_engagement_segment(service_count):
    if service_count <= 2:
        return "Low Engagement"
    elif service_count <= 5:
        return "Medium Engagement"
    else:
        return "High Engagement"

df["Engagement_Segment"] = df["Service_Count"].apply(create_engagement_segment)

# Check the distribution
print(df["Engagement_Segment"].value_counts())
```

```python
# Create a high-risk customer flag
# A customer is marked high-risk when multiple churn-related conditions are present

df["High_Risk_Flag"] = np.where(
    (
        (df["Contract"] == "Month-to-month").astype(int)
        + (df["tenure"] <= 12).astype(int)
        + (df["MonthlyCharges"] >= 80).astype(int)
        + (df["TechSupport"] == "No").astype(int)
        + (df["OnlineSecurity"] == "No").astype(int)
        + (df["Engagement_Segment"] == "Low Engagement").astype(int)
    ) >= 3,
    1,
    0
)

# Check distribution
print("High-Risk Customer Distribution:")
print(df["High_Risk_Flag"].value_counts())
```

```python
# Create monthly revenue at risk
# If the customer has churned, their MonthlyCharges are treated as lost monthly revenue.
# If the customer has not churned, revenue at risk is 0.

df["Monthly_Revenue_at_Risk"] = np.where(
    df["Churn_Flag"] == 1,
    df["MonthlyCharges"],
    0
)

# Create total historical value lost for churned customers
df["Customer_Value_Lost"] = np.where(
    df["Churn_Flag"] == 1,
    df["TotalCharges"],
    0
)

# Check sample results
print(
    df[
        [
            "customerID",
            "Churn",
            "MonthlyCharges",
            "TotalCharges",
            "Monthly_Revenue_at_Risk",
            "Customer_Value_Lost"
        ]
    ].head(10)
)
```

```python
# Calculate total customers
total_customers = df["customerID"].nunique()

# Calculate total churned customers
churned_customers = df.loc[
    df["Churn_Flag"] == 1,
    "customerID"
].nunique()

# Calculate overall churn rate
churn_rate = churned_customers / total_customers

print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers)
print("Overall Churn Rate:", round(churn_rate * 100, 2), "%")
```

```python
# Calculate churn rate by contract type
contract_churn = (
    df.groupby("Contract")
      .agg(
          Total_Customers=("customerID", "nunique"),
          Churned_Customers=("Churn_Flag", "sum")
      )
      .reset_index()
)

# Calculate churn percentage
contract_churn["Churn_Rate_%"] = (
    contract_churn["Churned_Customers"]
    / contract_churn["Total_Customers"]
    * 100
).round(2)

# Sort from highest churn rate to lowest
contract_churn = contract_churn.sort_values(
    "Churn_Rate_%",
    ascending=False
)

print(contract_churn)
```

```python
# Create a bar chart for churn rate by contract type
plt.figure(figsize=(8, 5))

plt.bar(
    contract_churn["Contract"],
    contract_churn["Churn_Rate_%"]
)

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=15)

plt.show()
```

```python
# Calculate churn rate by tenure group
tenure_churn = (
    df.groupby("Tenure_Group")
      .agg(
          Total_Customers=("customerID", "nunique"),
          Churned_Customers=("Churn_Flag", "sum")
      )
      .reset_index()
)

# Calculate churn percentage
tenure_churn["Churn_Rate_%"] = (
    tenure_churn["Churned_Customers"]
    / tenure_churn["Total_Customers"]
    * 100
).round(2)

# Custom order for tenure groups
tenure_order = [
    "0-12 Months",
    "13-24 Months",
    "25-48 Months",
    "49+ Months"
]

tenure_churn["Tenure_Group"] = pd.Categorical(
    tenure_churn["Tenure_Group"],
    categories=tenure_order,
    ordered=True
)

tenure_churn = tenure_churn.sort_values("Tenure_Group")

print(tenure_churn)
```

```python
# Create a bar chart for churn rate by tenure group
plt.figure(figsize=(8, 5))

plt.bar(
    tenure_churn["Tenure_Group"],
    tenure_churn["Churn_Rate_%"]
)

plt.title("Churn Rate by Tenure Group")
plt.xlabel("Tenure Group")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=15)

plt.show()
```

```python
# Calculate churn rate by internet service type
internet_churn = (
    df.groupby("InternetService")
      .agg(
          Total_Customers=("customerID", "nunique"),
          Churned_Customers=("Churn_Flag", "sum")
      )
      .reset_index()
)

# Calculate churn percentage
internet_churn["Churn_Rate_%"] = (
    internet_churn["Churned_Customers"]
    / internet_churn["Total_Customers"]
    * 100
).round(2)

# Sort from highest churn rate to lowest
internet_churn = internet_churn.sort_values(
    "Churn_Rate_%",
    ascending=False
)

print(internet_churn)
```

```python
# Create a bar chart for churn rate by internet service
plt.figure(figsize=(8, 5))

plt.bar(
    internet_churn["InternetService"],
    internet_churn["Churn_Rate_%"]
)

plt.title("Churn Rate by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")

plt.show()
```

```python
# Calculate churn rate by payment method
payment_churn = (
    df.groupby("PaymentMethod")
      .agg(
          Total_Customers=("customerID", "nunique"),
          Churned_Customers=("Churn_Flag", "sum")
      )
      .reset_index()
)

# Calculate churn percentage
payment_churn["Churn_Rate_%"] = (
    payment_churn["Churned_Customers"]
    / payment_churn["Total_Customers"]
    * 100
).round(2)

# Sort from highest churn rate to lowest
payment_churn = payment_churn.sort_values(
    "Churn_Rate_%",
    ascending=False
)

print(payment_churn)
```

```python
# Create a bar chart for churn rate by payment method
plt.figure(figsize=(10, 5))

plt.bar(
    payment_churn["PaymentMethod"],
    payment_churn["Churn_Rate_%"]
)

plt.title("Churn Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=25)

plt.show()
```

```python
# Calculate churn rate by monthly charge band
charge_churn = (
    df.groupby("Monthly_Charge_Band")
      .agg(
          Total_Customers=("customerID", "nunique"),
          Churned_Customers=("Churn_Flag", "sum")
      )
      .reset_index()
)

# Calculate churn percentage
charge_churn["Churn_Rate_%"] = (
    charge_churn["Churned_Customers"]
    / charge_churn["Total_Customers"]
    * 100
).round(2)

# Custom order
charge_order = [
    "Low Charges",
    "Medium Charges",
    "High Charges"
]

charge_churn["Monthly_Charge_Band"] = pd.Categorical(
    charge_churn["Monthly_Charge_Band"],
    categories=charge_order,
    ordered=True
)

charge_churn = charge_churn.sort_values("Monthly_Charge_Band")

print(charge_churn)
```

```python
# Create a bar chart for churn rate by monthly charge band
plt.figure(figsize=(8, 5))

plt.bar(
    charge_churn["Monthly_Charge_Band"],
    charge_churn["Churn_Rate_%"]
)

plt.title("Churn Rate by Monthly Charge Band")
plt.xlabel("Monthly Charge Band")
plt.ylabel("Churn Rate (%)")

plt.show()
```

```python
# Calculate churn rate by customer value segment
value_churn = (
    df.groupby("Customer_Value_Segment")
      .agg(
          Total_Customers=("customerID", "nunique"),
          Churned_Customers=("Churn_Flag", "sum")
      )
      .reset_index()
)

# Calculate churn percentage
value_churn["Churn_Rate_%"] = (
    value_churn["Churned_Customers"]
    / value_churn["Total_Customers"]
    * 100
).round(2)

# Set logical order
value_order = [
    "Low Value",
    "Medium Value",
    "High Value"
]

value_churn["Customer_Value_Segment"] = pd.Categorical(
    value_churn["Customer_Value_Segment"],
    categories=value_order,
    ordered=True
)

value_churn = value_churn.sort_values("Customer_Value_Segment")

print(value_churn)
```

```python
# Create a bar chart for churn rate by customer value segment
plt.figure(figsize=(8, 5))

plt.bar(
    value_churn["Customer_Value_Segment"],
    value_churn["Churn_Rate_%"]
)

plt.title("Churn Rate by Customer Value Segment")
plt.xlabel("Customer Value Segment")
plt.ylabel("Churn Rate (%)")

plt.show()
```

```python
# Calculate churn rate by engagement segment
engagement_churn = (
    df.groupby("Engagement_Segment")
      .agg(
          Total_Customers=("customerID", "nunique"),
          Churned_Customers=("Churn_Flag", "sum")
      )
      .reset_index()
)

# Calculate churn percentage
engagement_churn["Churn_Rate_%"] = (
    engagement_churn["Churned_Customers"]
    / engagement_churn["Total_Customers"]
    * 100
).round(2)

# Logical order
engagement_order = [
    "Low Engagement",
    "Medium Engagement",
    "High Engagement"
]

engagement_churn["Engagement_Segment"] = pd.Categorical(
    engagement_churn["Engagement_Segment"],
    categories=engagement_order,
    ordered=True
)

engagement_churn = engagement_churn.sort_values("Engagement_Segment")

print(engagement_churn)
```

```python
# Create a bar chart for churn rate by engagement segment
plt.figure(figsize=(8, 5))

plt.bar(
    engagement_churn["Engagement_Segment"],
    engagement_churn["Churn_Rate_%"]
)

plt.title("Churn Rate by Engagement Segment")
plt.xlabel("Engagement Segment")
plt.ylabel("Churn Rate (%)")

plt.show()
```

```python
# Compare high-risk flag with actual churn

risk_validation = (
    df.groupby("High_Risk_Flag")
      .agg(
          Total_Customers=("customerID", "nunique"),
          Churned_Customers=("Churn_Flag", "sum")
      )
      .reset_index()
)

# Calculate churn rate
risk_validation["Churn_Rate_%"] = (
    risk_validation["Churned_Customers"]
    / risk_validation["Total_Customers"]
    * 100
).round(2)

print(risk_validation)
```

```python
# Create bar chart for high-risk validation
plt.figure(figsize=(7, 5))

plt.bar(
    risk_validation["High_Risk_Flag"].astype(str),
    risk_validation["Churn_Rate_%"]
)

plt.title("Churn Rate by High-Risk Flag")
plt.xlabel("High-Risk Flag")
plt.ylabel("Churn Rate (%)")

plt.show()
```

```python
# Calculate overall financial impact of churn

total_monthly_revenue_at_risk = df["Monthly_Revenue_at_Risk"].sum()

total_customer_value_lost = df["Customer_Value_Lost"].sum()

print("Total Monthly Revenue at Risk:")
print(round(total_monthly_revenue_at_risk, 2))

print("\nTotal Customer Value Lost:")
print(round(total_customer_value_lost, 2))
```

```python
# Count high-value customers who have churned

high_value_churned = df[
    (df["Customer_Value_Segment"] == "High Value")
    & (df["Churn_Flag"] == 1)
]["customerID"].nunique()

print("High-Value Churned Customers:")
print(high_value_churned)
```

```python
# Final business insight summary

print("FINAL BUSINESS INSIGHTS")
print("-" * 50)

# 1. Overall churn rate
print(f"1. Overall Churn Rate: {churn_rate * 100:.2f}%")

# 2. Highest churn contract type
top_contract = contract_churn.iloc[0]
print(
    f"2. Highest Churn Contract: {top_contract['Contract']} "
    f"({top_contract['Churn_Rate_%']}%)"
)

# 3. Highest churn internet service
top_internet = internet_churn.iloc[0]
print(
    f"3. Highest Churn Internet Service: {top_internet['InternetService']} "
    f"({top_internet['Churn_Rate_%']}%)"
)

# 4. Highest churn payment method
top_payment = payment_churn.iloc[0]
print(
    f"4. Highest Churn Payment Method: {top_payment['PaymentMethod']} "
    f"({top_payment['Churn_Rate_%']}%)"
)

# 5. High-risk customer churn rate
high_risk_rate = risk_validation.loc[
    risk_validation["High_Risk_Flag"] == 1,
    "Churn_Rate_%"
].iloc[0]

print(f"5. High-Risk Customer Churn Rate: {high_risk_rate}%")

# 6. Monthly revenue at risk
print(
    f"6. Monthly Revenue at Risk: ${total_monthly_revenue_at_risk:,.2f}"
)

# 7. High-value churned customers
print(
    f"7. High-Value Churned Customers: {high_value_churned}"
)
```
