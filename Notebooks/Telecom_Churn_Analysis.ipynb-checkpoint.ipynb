```python
# Import chi-square test
from scipy.stats import chi2_contingency
```

```python
# Create a cross table between Contract and Churn
contract_churn_table = pd.crosstab(
    df["Contract"],
    df["Churn"]
)

# Show the table
contract_churn_table
```

```python
# Run chi-square test
chi2, p_value, dof, expected = chi2_contingency(
    contract_churn_table
)

# Show the results
print("Chi-Square Value:", round(chi2, 2))
print("P-Value:", p_value)
print("Degrees of Freedom:", dof)
```

```python
# Check if the relationship is statistically significant
if p_value < 0.05:
    print("Contract Type and Churn are significantly related.")
else:
    print("No significant relationship found.")
```
